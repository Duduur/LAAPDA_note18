## Passo 8 — Inicializar atributos com `__init__`

O método `__init__` é chamado durante a inicialização da instância. `self` referencia o objeto inicializado, enquanto `nome` e `ano` recebem os argumentos informados.

**Atenção:** nesta etapa redefinimos a classe `Musica`. Os objetos criados antes permanecem associados à definição anterior.

```python
class Musica:
    def __init__(self, nome, ano):
        self.nome = nome
        self.ano = ano

# Duas formas equivalentes de passar os argumentos:

# Argumentos nomeados (por palavra-chave)
# Obs.: A associação é realizada pelo nome do parâmetro, independentemente da ordem em que os argumentos nomeados são informados.
m1 = Musica(nome="Mina", ano=2026)
print(m1.nome, m1.ano)


# Argumentos posicionais
# Nesse formato, informamos apenas os valores, respeitando a ordem dos parâmetros definidos no método __init__.
m1 = Musica("Mina", 2026)
print(m1.nome, m1.ano)
```

### Teste — A ordem dos argumentos importa

```python
m2 = Musica(2026, "Mina")
print(m2.nome)
print(m2.ano)
```

## Passo 9 — Planejar os atributos da versão completa

Vamos evoluir a plataforma para armazenar mais informações e controlar reproduções.

| Atributo | Tipo esperado | Finalidade |
|---|---|---|
| `nome` | `str` | Nome da música |
| `cantor` | `str` | Cantor(a) |
| `genero` | `str` | Gênero musical |
| `duracao` | `int` | Duração em segundos |
| `ano` | `int` | Ano de lançamento da música |
| `reproducoes` | `int` | Contador de reproduções iniciadas |
| `tocando` | `bool` | Indica se a música está em reprodução |

Os cinco primeiros valores serão informados na criação do objeto. Os dois últimos começarão com valores padrão.

## Passo 10 — Criar a classe completa com atributos e métodos

Nesta versão, `__init__` inicializa os atributos. Os métodos `mostrar_*` exibem informações; `mostrar_detalhe()` reutiliza esses métodos; `reproduzir()` e `pausar()` modificam o estado da música.

**Importante:** o contador aumenta quando uma reprodução é **iniciada**, e não a cada chamada feita enquanto a música já está tocando.

```python
class Musica:
    def __init__(self, nome, cantor, genero, duracao, ano):
        self.nome = nome
        self.cantor = cantor
        self.genero = genero
        self.duracao = duracao
        self.ano = ano

        # Atributos de controle
        self.reproducoes = 0
        self.tocando = False
        self.favorito = False

    # Métodos para exibir informações
    def mostrar_nome(self):
        print(f"Nome da música: {self.nome}.")

    def mostrar_cantor(self):
        print(f"Cantor(a): {self.cantor}.")

    def mostrar_genero(self):
        print(f"Gênero: {self.genero}.")

    def mostrar_duracao(self):
        print(f"Duração: {self.duracao} segundos.")

    def mostrar_ano(self):
        print(f"Ano: {self.ano}.")

    # Método geral, que reutiliza outros métodos
    def mostrar_detalhe(self):
        self.mostrar_nome()
        self.mostrar_cantor()
        self.mostrar_genero()
        self.mostrar_duracao()
        self.mostrar_ano()
        print(f"Favorita: {"SIM" if self.favorito else "NÃO"}")
        print(f"Tocando: {"SIM" if self.tocando else "NÃO"}")
        print(f"Reproduções: {"SIM" if self.reproducoes > 0 else "NÃO"}")

    # Métodos que modificam o estado do objeto
    def reproduzir(self):
        if self.tocando == False:
            self.tocando = True
            self.reproducoes += 1
            print(f"Reproduzindo: {self.nome}")
        else:
            print("Essa música já está sendo tocada!")

    def pausar(self):
        if self.tocando == True:
            self.tocando = False
            print("Música pausada!")
        else:
            print("A música não está sendo reproduzida!")


    # ---------------------------------------------
    #Modulo de Alteração

    def alterar_nome(self, novo_nome):
        self._nome = novo_nome

    def alterar_cantor(self, novo_cantor):
        self._cantor = novo_cantor

    def alterar_genero(self, novo_genero):
        self._genero = novo_genero

    def alterar_duracao(self, novo_duracao):
        self._duracao = novo_duracao

    def alterar_ano(self, novo_ano):
            self.ano = novo_ano


    def favoritar(self):

        if self.favorito == False:
            self.favorito = True
            print(f"{self.nome} foi adicionado aos favoritos!")
        else:
            print(f"{self.nome} já está nos favoritos!")

    def desfavoritar(self):
        if self.favorito == True:
            self.favorito = False
            print(f"{self.nome} foi removido dos favoritos!")
        else:
            print(f"{self.nome} não está nos favoritos!")
```

```python
class Musica:
    def __init__(self, nome, cantor, genero, duracao, ano):

        #Validação
        if not isinstance(nome, str)or not nome.strip():
            raise ValueError("O nome da música deve ser uma string.")
        if not isinstance(cantor, str) or not cantor.strip():
            raise ValueError("O nome do cantor deve ser uma string.")
        if not isinstance(genero, str) or not genero.strip():
            raise ValueError("O gênero da música deve ser uma string.")
        if type(duracao) is not int or duracao <= 0:
            raise ValueError("A duração da música deve ser um número positivo.")
        if not isinstance(ano, int):
            raise ValueError("O ano da música deve ser um número inteiro.")


        self._nome = nome
        self._cantor = cantor
        self._genero = genero
        self._duracao = duracao
        self._ano = ano

        # Atributos de controle
        self.reproducoes = 0
        self.tocando = False
        self.favorito = False

    # Métodos para exibir informações

    # ======================================================================

    def __str__(self):

        minutos = self._duracao // 60
        segundos = self._duracao % 60
        
        return (f"{self._nome} - {self._cantor} - {minutos}:{segundos:02d}")

    # ======================================================================

    
    def mostrar_nome(self):
        if not isinstance(self.nome, str) or not self.nome.strip():
            raise ValueError("O nome da música deve ser uma string.")
        print(f"Nome da música: {self.nome}.")

    def mostrar_cantor(self):
        if not isinstance(self.cantor, str) or not self.cantor.strip():
            raise ValueError("O nome do cantor deve ser uma string.")
        print(f"Cantor(a): {self.cantor}.")

    def mostrar_genero(self):
        if not isinstance(self.genero, str) or not self.genero.strip():
            raise ValueError("O gênero da música deve ser uma string.")
        print(f"Gênero: {self.genero}.")

    def mostrar_duracao(self):
        if type(self.duracao) is not int or self.duracao <= 0:
                    raise ValueError("A duração da música deve ser um número positivo.")
        print(f"Duração: {self.duracao} segundos.")

    def mostrar_ano(self):
        if type(self.ano) is not int or self.ano <= 0:
            raise ValueError("O ano da música deve ser um número inteiro.")
        print(f"Ano: {self.ano}.")

    # Método geral, que reutiliza outros métodos
    def mostrar_detalhe(self):
        self.mostrar_nome()
        self.mostrar_cantor()
        self.mostrar_genero()
        self.mostrar_duracao()
        self.mostrar_ano()
        
        print(f"Favorita: {"SIM" if self.favorito else "NÃO"}")
        print(f"Tocando: {"SIM" if self.tocando else "NÃO"}")
        print(f"Reproduções: {"SIM" if self.reproducoes > 0 else "NÃO"}")

    # Métodos que modificam o estado do objeto
    def reproduzir(self):
        if self.tocando == False:
            self.tocando = True
            self.reproducoes += 1
            print(f"Reproduzindo: {self._nome}")
        else:
            print("Essa música já está sendo tocada!")

    def pausar(self):
        if self.tocando == True:
            self.tocando = False
            print("Música pausada!")
        else:
            print("A música não está sendo reproduzida!")


    # ---------------------------------------------
    #Modulo de Alteração

    def alterar_nome(self, novo_nome):
        if not isinstance(novo_nome, str) or not novo_nome.strip():
            raise ValueError("O nome da música deve ser uma string.")
        self.nome = novo_nome

    def alterar_cantor(self, novo_cantor):
        if not isinstance(novo_cantor, str) or not novo_cantor.strip():
            raise ValueError("O nome do cantor deve ser uma string.")
        self.cantor = novo_cantor

    def alterar_genero(self, novo_genero):
        if not isinstance(novo_genero, str) or not novo_genero.strip():
            raise ValueError("O gênero da música deve ser uma string.")
        self.genero = novo_genero

    def alterar_duracao(self, novo_duracao):
        if type(novo_duracao) is not int or novo_duracao <= 0:
            raise ValueError("A duração da música deve ser um número positivo.")
        self.duracao = novo_duracao

    def alterar_ano(self, novo_ano):
        if type(novo_ano) is not int or novo_ano <= 0:
            raise ValueError("O ano da música deve ser um número inteiro.")
        self.ano = novo_ano


    def favoritar(self):

        if self.favorito == False:
            self.favorito = True
            print(f"{self._nome} foi adicionado aos favoritos!")
        else:
            print(f"{self.nome} já está nos favoritos!")

    def desfavoritar(self):
        if self.favorito == True:
            self.favorito = False
            print(f"{self._nome} foi removido dos favoritos!")
        else:
            print(f"{self._nome} não está nos favoritos!")


## ---------------------------------------------------------------

# Propriedade @property

    @property
    def nome(self):
        return self._nome

    @property
    def cantor(self):
        return self._cantor 

    @property
    def genero(self):
        return self._genero

    @property
    def duracao(self):
        return self._duracao

    @property
    def ano(self):
        return self._ano
    
    
```

## Síntese da aula

- **Classe:** define uma estrutura e comportamentos.
- **Objeto/instância:** entidade criada a partir da classe.
- **Atributo:** informação associada ao objeto.
- **`__init__`:** inicializa atributos da instância.
- **`self`:** referencia a instância em um método.
- **Método:** função definida na classe que pode consultar ou alterar o estado de um objeto.

### Situação-problema

Uma plataforma de músicas precisa permitir que a duração de uma música já cadastrada seja corrigida. Entretanto, o sistema não deve aceitar durações inválidas, como valores iguais ou inferiores a zero.

**Desafio:** implemente um método chamado `atualizar_duracao(self, nova_duracao)` na classe `Musica`, permitindo atualizar a duração de uma música existente.

1. Criar o método `atualizar_duracao(self, nova_duracao)` dentro da classe `Musica`.

5. Caso contrário, impedir a atualização e apresentar uma mensagem de erro.
6. Testar o método com valores válidos e inválidos.
7. Consultar o atributo `duracao` após cada teste para verificar se a atualização ocorreu corretamente.



### Atenção

**Se a duração informada for inválida, o objeto deverá manter a duração que possuía anteriormente.**
