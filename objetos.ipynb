```python
class Musica:
    def __init__(self, nome, cantor,genero,duracao,ano):
        self.nome = nome
        self.cantor = cantor
        self.genero = genero
        self.duracao = duracao
        self.ano = ano

        #Atributos de controle
        self.reproducoes = 0
        self.tocando = False

    # Padroize os nomes das função
    # Evite colocar os nomes da função, com o mesmo nome do atributo

    def mostrar_nome(self):
        print(f"Nome da música {self.nome}")
    def mostrar_cantor(self):
        print(f"Cantor(a): {self.cantor}")
    def mostrar_genero(self):
        print(f"Gênero: {self.genero}")
    def mostrar_duracao(self):
        print(f"Duração: {self.duracao}")
    def mostrar_ano(self):
        print(f"Ano: {self.ano}")

# ========================================= # 
    def mostrar_detalhes(self):
        self.mostrar_nome
        self.mostrar_cantor
        self.mostrar_genero
        self.mostrar_duracao
        self.mostrar_ano

# =========================================== #

    def reproduzir(self):
        if self.tocando == False:

            self.reproducoes +=1
            self.tocando = True
            print(f"Reproduzindo: {self.nome}")
        else:
            print(f"Essa musica já está sendo executada")


# =========================================== #
    def pausar(self):
        if self.tocando != True:
            self.tocando = False
            print(f"Música pausada!")
        else:
            print("Musica nâ está tocando")
```

```python
m1 = Musica("Trap de Cria 2", "Kyan", "Funk",  73, 2022 )

m1.mostrar_nome()
m1.mostrar_cantor()
m1.mostrar_genero()
m1.mostrar_duracao()
m1.mostrar_ano()


print("\nInformaçôes detalhadas")
m1.mostrar_detalhes()
```

```python
  
m1.reproduzir()
print(m1.reproducoes)
```
