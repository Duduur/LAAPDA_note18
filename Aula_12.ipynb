```python
class Musica:
    def __init__(self, nome, cantor, genero, duracao, ano):
        self.nome = nome
        self.cantor = cantor
        self.genero = genero
        self.duracao = duracao
        self.ano = ano

        # Atributos de controle
        self.reproducoes = 0
        self.tocando = False

    # Métodos para exibir informações
    def mostrar_nome(self):
        print(f"Nome da música: {self.nome}.")

    def mostrar_cantor(self):
        print(f"Cantor(a): {self.cantor}.")

    def mostrar_genero(self):
        print(f"Gênero: {self.genero}.")

    def mostrar_duracao(self):
        print(f"Duração: {self.duracao} segundos.")

    def mostrar_ano(self):
        print(f"Ano: {self.ano}.")

    # Método geral, que reutiliza outros métodos
    def mostrar_detalhe(self):
        self.mostrar_nome()
        self.mostrar_cantor()
        self.mostrar_genero()
        self.mostrar_duracao()
        self.mostrar_ano()

    # Métodos que modificam o estado do objeto
    def reproduzir(self):
        if self.tocando == False:
            self.tocando = True
            self.reproducoes += 1
            print(f"Reproduzindo: {self.nome}")
        else:
            print("Essa música já está sendo tocada!")

    def pausar(self):
        if self.tocando == True:
            self.tocando = False
            print("Música pausada!")
        else:
            print("A música não está sendo reproduzida!")


    # ---------------------------------------------
    #Modulo de Alteração

    def alterar_nome(self, novo_nome):
        self.nome = novo_nome

    def alterar_cantor(self, novo_cantor):
        self.cantor = novo_cantor

    def alterar_genero(self, novo_genero):
        self.genero = novo_genero

    def alterar_duracao(self, novo_duracao):
        self.duracao = novo_duracao

    def alterar_ano(self, novo_ano):
            self.ano = novo_ano
```
